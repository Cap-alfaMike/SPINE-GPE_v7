```python

from pathlib import Path
from google.colab import files
import hashlib, shutil, zipfile

PACKAGE_NAME = "SPINE_GPEv7_PHASE1_EVIDENCE_FOUNDATION_PACKAGE_v1.0.0.zip"
PACKAGE_ZIP = Path("/content") / PACKAGE_NAME
ROOT = Path("/content/drive/MyDrive/aCidadeAlgoritmica/SPINE-GPEv7")
INSTALL_DIR = ROOT / "scripts" / "phase1_evidence_foundation_v100"

if not PACKAGE_ZIP.is_file():
    print(f"Selecione {PACKAGE_NAME}")
    uploaded = files.upload()
    assert PACKAGE_NAME in uploaded, list(uploaded)

def sha256_file(path, chunk_size=8*1024*1024):
    h=hashlib.sha256()
    with open(path,'rb') as f:
        for chunk in iter(lambda:f.read(chunk_size), b''):
            h.update(chunk)
    return h.hexdigest()

print("Package:", PACKAGE_ZIP)
print("SHA-256:", sha256_file(PACKAGE_ZIP))

if INSTALL_DIR.exists():
    shutil.rmtree(INSTALL_DIR)
INSTALL_DIR.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(PACKAGE_ZIP) as zf:
    zf.extractall(INSTALL_DIR)
print("Installed:", INSTALL_DIR)
print("Files:", len(list(INSTALL_DIR.rglob('*'))))
```

```python

import subprocess, sys
ENGINE = INSTALL_DIR / "SPINE_GPEv7_PHASE1_EVIDENCE_FOUNDATION_v1.0.0.py"
cmd=[sys.executable,str(ENGINE),'--root',str(ROOT),'--mode','audit','--run-id','phase1_evidence_foundation_audit_v100','--strict']
print(' '.join(cmd))
result=subprocess.run(cmd,text=True)
print('Audit exit code:',result.returncode)
assert result.returncode==0
```

```python

cmd=[sys.executable,str(ENGINE),'--root',str(ROOT),'--mode','build','--run-id','phase1_evidence_foundation_final_v100','--strict']
print(' '.join(cmd))
result=subprocess.run(cmd,text=True)
print('Build exit code:',result.returncode)
assert result.returncode==0
```
