```python
import importlib
import subprocess
import sys

# Pacotes realmente necessários que podem não vir completos no Colab.
# Os valores são: módulo a importar -> pacote no PyPI.
DEPENDENCIAS = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scipy": "scipy",
    "pyarrow": "pyarrow>=17,<22",
    "requests": "requests",
    "urllib3": "urllib3",
    "bs4": "beautifulsoup4>=4.12,<5",
    "openpyxl": "openpyxl>=3.1,<4",
    "xlrd": "xlrd>=2.0,<3",
    "lxml": "lxml>=5,<7",
}

faltantes = []

for modulo, pacote in DEPENDENCIAS.items():
    try:
        importlib.import_module(modulo)
        print(f"OK       {modulo}")
    except ImportError:
        print(f"FALTANDO {modulo} -> {pacote}")
        faltantes.append(pacote)

if faltantes:
    print("\nInstalando:", faltantes)

    resultado = subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "--no-cache-dir",
            "--prefer-binary",
            "--retries",
            "10",
            "--timeout",
            "120",
            *faltantes,
        ],
        text=True,
        capture_output=True,
        check=False,
    )

    print("\nSTDOUT:")
    print(resultado.stdout[-10000:])

    print("\nSTDERR:")
    print(resultado.stderr[-10000:])

    if resultado.returncode != 0:
        raise RuntimeError(
            f"Falha na instalação. Exit code: {resultado.returncode}"
        )
else:
    print("\nTodas as dependências já estão disponíveis.")
```

```python
from pathlib import Path
import pandas as pd

ROOT = Path(
    "/content/drive/MyDrive/aCidadeAlgoritmica/SPINE-GPEv7"
)

BASE = ROOT / "03_processed/10_pnadc_certified"

for year in (2022, 2024):
    path = BASE / f"certified_pnadc_platform_{year}.parquet"

    df = pd.read_parquet(
        path,
        columns=[
            "SD14001",
            "S140093",
            "survey_weight",
        ],
    )

    sd = df["SD14001"].astype("string").str.strip()
    entrega = df["S140093"].astype("string").str.strip()
    peso = pd.to_numeric(df["survey_weight"], errors="coerce").fillna(0)

    uso_bruto_entrega = peso[entrega.eq("1")].sum()

    entrega_plataformizada_oficial = peso[
        sd.eq("1") & entrega.eq("1")
    ].sum()

    excluidos_pela_regra_oficial = (
        uso_bruto_entrega
        - entrega_plataformizada_oficial
    )

    print(f"\n{year}")
    print(
        "S140093 == 1:",
        f"{uso_bruto_entrega:,.0f}",
    )
    print(
        "SD14001 == 1 & S140093 == 1:",
        f"{entrega_plataformizada_oficial:,.0f}",
    )
    print(
        "Diferença:",
        f"{excluidos_pela_regra_oficial:,.0f}",
    )
```
